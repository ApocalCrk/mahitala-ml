```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis
```

```python
df = pd.read_csv('home/noturminesv/projects/gis-ml/dataset/dataset_yogyakarta_augmented.csv')

df_numeric = df.drop(columns=['label'])

for col in df_numeric.columns:
    if not pd.api.types.is_numeric_dtype(df_numeric[col]):
        df_numeric = df_numeric.drop(columns=[col])

df_numeric
```

```python
df_numeric.hist(bins=10, figsize=(12, 8))
plt.suptitle('Histogram of Features')
plt.show()
```

```python
for col in df_numeric.columns:
    print(f"Feature: {col}")
    print(f"  Mean: {df_numeric[col].mean()}")
    print(f"  Std Dev: {df_numeric[col].std()}")
    print(f"  Skewness: {skew(df_numeric[col])}")
    print(f"  Kurtosis: {kurtosis(df_numeric[col])}")
    print()
```
